## Abstract

This study investigates whether disparities exist in legal penalties based on defendant characteristics, including gender, race, and county. Using regression models, ANOVA, non-parametric tests, and chi-square tests, we evaluated potential sources of bias in sentencing outcomes. Our results indicate that gender does not have a statistically significant effect on penalties (p > 0.05). However, we find that county, race, and specific judges within certain counties are associated with differences in penalties (p < 0.05). These findings suggest that structural and regional factors may influence sentencing outcomes. Limitations in data and methodology are discussed in detail later in the report.

## How problems are analyzed

## Contexts before reading the text below:

**Association (used repeatedly across models):** A relationship between two variables where one tends to change alongside the other, but it doesn't mean that one causes the other to change, it could be from other factors. For example, if race had an association with fine amounts, it doesn't mean which race you are, your fines will be determined as this specific amount because the change could've been caused by which judge from which county did the sentencing.

**Baseline:** A group that every other group compares to. For example, if White defendants were a baseline, then all other races are compared to White defendants for this portion of analysis.

**Binary outcomes:** The results can only be one of two things. For example, is the person guilty or not guilty, did they receive the jail penalty or didn't, etc.

**Coefficients:** Numbers that the model calculates which shows how strong and in which direction the relationship is.

**Continuous outcomes:** The results can be any numerical value, including decimals. For example, length of probation months, dollar amount of fines, etc.

**Categorical variables:** Data that have no numeric value, but fits into groups. For example, race, gender, county, judge names, etc.

**Direction (positive or negative association):** The relationship between two variables. If one variable make the other go up, the direction is positive. However, if one variable make the other variable go down, the direction is negative.

**Discrete outcomes:** The results are unique and countable. For example, number of males, number of White defendants, etc. There can't be a number of 2.5 males or 3.79 White defendants.

**Expected counts per cell:** In a table, how many counts would you expect to be in each boxes if there were no relationship between the two groups or variables.

**Effect size and "Magnitude of differences":** How big the difference is between the two groups.

**Hypothesis testing:** A formal way of asking if the pattern is real or just randomness.

**MLE (Maximum Likelihood Estimator):** The method used to find the best-fitting coefficients for the patterns in our data.

**Normal distribution:** Data is shaped like a bell-curve, most values cluster around the middle with extreme values on either sides. This matters because a lot of statistical tests assume your data to look like this and if it doesn't our results wouldn't be reliable.

**Odds ratio:** How much more or less likely an outcome for one group against another group. For example, if males had an odds ratio of 1.4, it means that the outcome happening for males are 1.4 times the odds for the other group that is compared against male.

**Ordered categories:** Groups that have natural ranking. For example, if you get jail as a punishment, most people would say receiving a $5 dollar fine would be lower punishment than getting jailed.

**"Order matters" (in ordinal regression):** The categories aren't just different, they represent levels of something. For example, probation is less severe than jail, so the "order matters".

**Pairwise differences:** Comparing every possible pair from of groups against each other. For example, White vs Black defendants, Black vs Hispanic defendants, and White vs Hispanic defendants.

**Penalty score:** A numeric scale we created to rank penalties from least to most severe. (We asked our sponsor before creating this to make sure our rank is correct)

**"Robust with large samples":** The model results stay reliable even when the data isn't perfectly clean or you have a lot of data.

**Sample size per group:** How many individual cases fall into each group. For example, number of White defendants, number of male defendants, etc.

**Statistical significance (p < 0.05):** If a statistic has a statistical significance less than 0.05 or "p < 0.05", it means that the result didn't happen due to randomness, but rather a real and meaningful finding.

**Skewed data:** Values that are piled up on one end. For example, most people would get fined around \$1,000, but there are considerable amount of people who were fined around \$3,000 which caused our data to be "skewed".

**"Spreadability" (in Mann-Whitney):** How much the values in a group vary from each other. For example, if most male sentences are from 0 to 30 days and female sentences are from 0 to 90 days, female sentences have higher "spreadability".

## Our Models

**Chi-square tests:** Used to test binary outcomes, works for categorical variables, finding associations, discrete outcomes

Chi-square tests works well because it doesn't assume normal distribution, work well with large-datasets, and also uses "expected counts per cell" for calculating. Which basically means that, as long as your "expected count" is more than or equal to 5, then the results would be valid. However, this doesn't measure direction or the strength of the difference. So, we investigated further using logistic regression.

**Binary logistic regression:** Used to test binary outcomes, works for categorical variables, for finding associations, finding effect size, direction, significance, discrete outcomes

Binary logistic regression works well because after finding if the variables has association from chi-square tests, we can use Binary Logistic Regression to find the direction (if the association is positive or negative) and the relevance of our analysis.

**Ordinal Logistic Regression:** Used to test penalty severities outcomes (outcomes with 3 or more ordered categories), works for categorical variables, for finding associations, statistical significance, effect size, discrete outcomes

Ordinal Logistic Regression is very similar to Binary Logistic Regression, but the ordinal logistic regression can analyze data when order matters. For example, we believe that people who receive Jail had harsher punishment than people who received Probation. This is ordering, and ordinal logistic regression is a good model for analyzing this.

**Sample size concerns for logistic regressions:** For logistic regression, sample size is not a big issue due to how the probabilities are calculated. Logistic regression uses MLE to estimate the odds ratio which doesn't require equal groups of sample size.

**Maximum Likelihood Estimator** tries to find coefficients that maximizes the probability of seeing the penalties that occured in this dataset. It uses individual observations to estimate probabilities. Having more males than females, just means we have more information on males, but doesn't mean that the dataset is bias. (just need enough sample size for each group, minimum 100 sample size per group)

**Mann-Whitney U Test:** Used for comparing two groups (in this case gender) against a penalty, hypothesis testing, discrete outcomes

Mann-Whitney U test works well because it doesn't assume normal distribution and equal spreadability, it's perfect for skewed data or messy data. It answers the question that if we randomly pick one male and one female case, what is the probability that the male has a higher penalty score? (penalty score, from our "assumptions") As long as each group has a reasonable number (30+ samples per each group), the model is valid.

**Kruskal Wallis test:** Used for comparing multiple groups (in this case, race and counties) against a penalty, hypothesis testing, discrete outcomes

Kruskal Wallis test is an extension of Mann-whitney U to multiple groups. It's perfect for messy, uneven data, and doesn't require equal group sizes. As long as each group has a reasonable number (30+ samples per each group), the model is valid.

**OLS Regression:** Used for analyzing continuous outcomes, magnitude of differences, interpretability

OLS Regression works well because the model is robust with large samples, it utilizes MLE like other regression models, it shows the magnitude of our outcomes, and is a good model for analyzing continuous outcomes. One caution with this model is that when your data is very skewed, it can mess with the analysis. Our solution for this is to apply log on the outcome, and interpret it carefully.

**Tukey HSD:** Used after a significant OLS regression result, used to explore and compare if each group pairs are significantly difference from each other

While OLS regression compares each group against a baseline, it doesn't capture all pairwise differences. Tukey HSD was applied to pinpoint the other significant pairs that weren't represented in OLS regression results.

**Heatmap:** Used for showing patterns, directions and magnitudes of difference — validated statistically by ANOVA, no minimum sample size requirement, but larger sample size = more reliability

Heatmap is used to identify the magnitude and direction of each variable in the group (in this case, judges in each county). It gives us a visual presentation of the difference between each variable. However, to test if these differences are significant or not, we need other tests. In this case, we picked ANOVA test.

**ANOVA:** Used for comparing average of a variable across multiple groups, gives us statistical proof

ANOVA is used alongside the heatmap to statistically validate the visual patterns observed. Specifically, it tests whether the average penalty scores differ significantly across judges within each county. A significant ANOVA result confirms that the differences seen in the heatmap are unlikely to be due to chance, providing formal statistical evidence to support the visual findings.

```python
# Copy the df_selected_all column
df_selected_new = df_selected_all.copy()

# Create new columns to track each penalty that was given to the defendants
df_selected_new["received_fine"] = df_selected_new["penalties"].str.contains("Fine", case=False, na=False).astype(int)
df_selected_new["received_probation"] = df_selected_new["penalties"].str.contains("Probation", case=False, na=False).astype(int)
df_selected_new["received_treatment"] = df_selected_new["penalties"].str.contains(r"Treatment|Therapy|Mental Health|Substance|Assessment", case=False, na=False).astype(int)
df_selected_new["received_jail"] = df_selected_new["penalties"].str.contains("Jail", case=False, na=False).astype(int)
df_selected_new["received_community_service"] = df_selected_new["penalties"].str.contains("Community", case=False, na=False).astype(int)
df_selected_new["received_vip"] = df_selected_new["penalties"].str.contains("VIP", case=False, na=False).astype(int)
df_selected_new["received_alcohol_eval"] = df_selected_new["penalties"].str.contains("Alcohol", case=False, na=False).astype(int)
df_selected_new["received_restitution"] = df_selected_new["penalties"].str.contains("Restitution", case=False, na=False).astype(int)
df_selected_new["received_detention"] = df_selected_new["penalties"].str.contains(r"Detention|house arrest|home detention", case=False, na=False).astype(int)
df_selected_new["received_DUI"] = df_selected_new["penalties"].str.contains("DUI", case=False, na=False).astype(int)
df_selected_new["received_work_release"] = df_selected_new["penalties"].str.contains("Work release", case=False, na=False).astype(int)
df_selected_new["received_monitoring"] = df_selected_new["penalties"].str.contains(r"SCRAM|Interlock|Monitored|UA|MOPS|day-reporting", case=False, na=False).astype(int)
```

**Visualization:** Received Penalties by Gender (%)


```python
# Suppose received_cols is a list of all your received_* columns
received_cols = [
    "received_fine",
    "received_probation",
    "received_treatment",
    "received_jail",
    "received_community_service",
    "received_vip",
    "received_alcohol_eval",
    "received_restitution",
    "received_detention",
    "received_DUI",
    "received_work_release",
    "received_monitoring"
]

# Create an empty list
results_chi = []

# Loop through each penalty and find the chi-square result
for col in received_cols:
    # Build contingency table
    table = pd.crosstab(df_selected_new["gender"], df_selected_new[col])

    # Run chi-square
    chi2, p, dof, ex = chi2_contingency(table)

    results_chi.append({
        "Penalty": col,
        "Chi2": chi2,
        "p-value": p
    })

# Store the chi-square results in a dataframe
df_chi2_results = pd.DataFrame(results_chi)

# Show the dataframe
print(df_chi2_results)
```

**Interpretation:**

Chi-square tests of independence were conducted to examine whether gender is associated with each type of penalties. Across all 12 penalty types, p-values ranged from 0.058 to 0.962, with none reaching the statistical significance threshold of 0.05.

Therefore, we fail to reject the null hypothesis for all penalties and find no statistically significant evidence of a gender & penalty association in this dataset. Notably, monitoring produced the lowest p-value (p-value = 0.058), suggesting a potential trend that could be explored further with more samples.

**Key Takeaway:**

The Chi-square tests revealed no significant association between gender and any of the penalty types.

**Visualization:** Chi-square Visualization (Gender)

```python
# Make a copy of the df_selected_new dataframe
df_clean = df_selected_new.copy()

# Create a new column that stores the gender as binary value
df_clean["gender_binary"] = df_clean["gender"].map({ "Male": 0, "Female": 1 })

# Create an empty list
results = []

# Adds a column of all 1s to your data before running a regression
X = sm.add_constant(df_clean["gender_binary"])

# Loop through and calculate the logistic regression value for each penalty
for col in received_cols:

    y = df_clean[col].astype(int)

    model = sm.Logit(y, X).fit(disp=0)  # disp=0 suppresses output

    coef = model.params["gender_binary"]
    pval = model.pvalues["gender_binary"]
    odds_ratio = np.exp(coef)

    results.append({
        "Penalty": col,
        "Coefficient": coef,
        "Odds Ratio": odds_ratio,
        "p-value": pval
    })

# Store the results in a data frame
results_df = pd.DataFrame(results)

# Show the data frame
print(results_df)
```

**Interpretation:**

Logistic regression was conducted to examine the association between gender and each penalty type. Across all penalties, no statistically significant associations were found. The most noteworthy result was for monitoring (which includes electric home monitoring, monitored sobriety, and similar conditions), where females had approximately 22% higher odds of receiving monitoring compared to males (Odds Ratio = 1.22, p-value = 0.052).

However, this result narrowly misses the statistical significance threshold of 0.05 and should therefore be interpreted cautiously.

**Key Takeaway:**

The Logistic Regression found no significant association between gender and any penalty types.

```python
# Ordered Logistic Regression of gender + county and penalty severity

# Create our severity map
# (1 = least severe, 4 = most severe)
severity_map = {
    "Probation": 1,
    "Detention": 2,
    "Work Release": 3,
    "Jail": 4
}

# Make a copy of the cleaned dataset to avoid modifying the original
df_gender_county_penalty_score = df_gender_county_penalty.copy()

# Convert penalty severity into a numeric ordinal variable using the mapping
df_gender_county_penalty_score["penalty_score"] = df_gender_county_penalty_score["penalty_severity"].map(severity_map)

# Remove any rows where mapping failed
df_gender_county_penalty_score = df_gender_county_penalty_score[df_gender_county_penalty_score["penalty_score"] != 0]

# Select predictor variables
X = df_gender_county_penalty_score[["gender", "county"]]

# Convert categorical variables into dummy variables
X = pd.get_dummies(X, drop_first=True)

# Define the ordinal logistic regression model
model = OrderedModel(
    df_gender_county_penalty_score["penalty_score"],
    X,
    distr="logit"
)

result = model.fit(method="bfgs")

# Print regression results (coefficients, p-values, etc.)
print(result.summary())

# Convert log-odds coefficients into odds ratios for easier interpretation
odds_ratios = np.exp(result.params)
print("\nOdds Ratios:")
print(odds_ratios)
```

**Interpretation:**

The ordered logistic regression suggests that county of residence influences penalty severity, while gender shows minimal effect. Being male has a near-zero effect on penalties (coef = 0.015; odds ratio = 1.015), indicating virtually no difference from females.  Among counties, Boulder (coef = 0.709; OR = 2.03) and Larimer (coef = 0.630; OR = 1.88) are associated with higher odds of more severe penalties, whereas El Paso (coef = -0.947; OR = 0.39), La Plata (coef = -1.619; OR = 0.20), and Pueblo (coef = -1.455; OR = 0.23) show lower odds. These values highlight that geography has a larger impact on penalty outcomes than gender, and there is clear variation in severity across counties.

**Key Takeaway:**

The model shows that county plays a much bigger role than gender in determining penalty severity, with noticeable differences across counties.

This investigation includes the suspended fines and jail. I didn't include suspended probations because that didn't exist in this dataset.

```python
# Extract sentencing components from text
# Create a copy of the race-filtered dataset for quantitative sentencing analysis
df_quantify = df_selected_newrace.copy()

# Define function to extract numeric sentencing values from messy sentencing text
def extract_sentencing_components(text):
    """
    Extract numeric values for:
    - Jail served
    - Jail suspended
    - Fine served
    - Fine suspended
    - Probation months
    Handles messy text and multiple formats.
    """
    # Ensure text is string
    if not isinstance(text, str):
        text = ""

    # Jail served (exclude suspended)
    # Matches: "10 days jail", "435 days imposed", "10 days to be served"
    match_jail_served = re.findall(
        r'(\d+)\s*day[s]?\s*(?:jail|imposed|to be served)(?!.*suspended)',
        text, flags=re.IGNORECASE
    )
    jail_served = sum(int(x) for x in match_jail_served) if match_jail_served else 0

    # Jail suspended
    # Matches: "365 days jail suspended", "365 days suspended"
    match_jail_suspended = re.findall(
        r'(\d+)\s*day[s]?\s*(?:jail\s*time\s*suspended|suspended)',
        text, flags=re.IGNORECASE
    )
    jail_suspended = sum(int(x) for x in match_jail_suspended) if match_jail_suspended else 0

    # Fine served
    # Matches: "$325 fine" but NOT "fine suspended"
    match_fine = re.findall(
        r'\$?(\d+)\s*fine(?!\s*suspended)', text, flags=re.IGNORECASE
    )
    fine_amount = sum(int(x) for x in match_fine) if match_fine else 0

    # Fine suspended
    # Matches: "$150 fine suspended"
    match_fine_suspended = re.findall(
        r'\$?(\d+)\s*fine\s*suspended', text, flags=re.IGNORECASE
    )
    fine_suspended = sum(int(x) for x in match_fine_suspended) if match_fine_suspended else 0

    # Probation months
    # Matches: "12 months probation"
    match_prob = re.findall(
        r'(\d+)\s*month[s]?\s*(?=.*probation)', text, flags=re.IGNORECASE
    )
    prob_months = sum(int(x) for x in match_prob) if match_prob else 0

    return pd.Series([jail_served, jail_suspended, fine_amount, fine_suspended, prob_months])

# Apply to dataframe
df_quantify[["jail_served", "jail_suspended", "fines",
             "fines_suspended", "probs"]] = df_quantify["sentencing"].apply(
    extract_sentencing_components
)

# Ensure numeric
cols = ["jail_served", "jail_suspended", "fines", "fines_suspended", "probs"]
df_quantify[cols] = df_quantify[cols].apply(pd.to_numeric, errors="coerce")

# Optional: create total columns for modeling
df_quantify["total_jail"] = df_quantify["jail_served"] + df_quantify["jail_suspended"]
df_quantify["total_fine"] = df_quantify["fines"] + df_quantify["fines_suspended"]
```

```python
def clean_name(name):
    if pd.isna(name):
        return None

    name = name.lower().strip()

    # remove "judge"
    name = re.sub(r'\bjudge\b', '', name)

    # remove extra stuff in parentheses
    name = re.sub(r'\(.*?\)', '', name)

    # remove extra whitespace
    name = re.sub(r'\s+', ' ', name)

    return name.strip()
```

```python
df_judge = df_judge[df_judge["obs_judge_name"] != "Judge not listed (list in next field along with courtroom number)"]
# df_judge = df_judge[df_judge["obs_judge_name"].str.strip() != ""]
```

```python
# Filter dataset to include only selected counties of interest
counties = [
    "Denver", "El Paso", "Douglas", "Arapahoe", "Boulder",
    "Mesa", "Larimer", "Adams", "Weld", "Broomfield",
    "Pueblo", "Jefferson"
]

df_subset = df_judge[df_judge["county"].isin(counties)]
```

```python
# Create groups of fine amounts for each judge (required input format for ANOVA)
groups = [
    group["fines"].dropna()
    for _, group in df_subset.groupby("obs_judge_name")
]

# Apply one-way ANOVA to test whether mean fines differ across judges
f_stat, p_value = stats.f_oneway(*groups)

print(p_value)
```

```python
# Ensure fines are numeric and remove missing values before analysis

df_subset = df_subset.dropna(subset=["fines"])
df_subset["fines"].value_counts()
df_subset["fines"] = pd.to_numeric(df_subset["fines"], errors="coerce")
df_subset["obs_judge_name"] = df_subset["obs_judge_name"].astype(str)
```

```python
# Run ANOVA within each county to test whether jail time differs across judges
results = []

for county in df_subset["county"].unique():

    df_c = df_subset[df_subset["county"] == county]

    # Keep only judges with sufficient cases to ensure reliable comparison
    valid = df_c["obs_judge_name"].value_counts()
    valid = valid[valid >= 20].index
    df_c = df_c[df_c["obs_judge_name"].isin(valid)]

    # Skip if not enough judges
    # Skip counties with fewer than two judges
    if df_c["obs_judge_name"].nunique() < 2:
        continue

    # Create groups of jail time for each judge (required for ANOVA)
    groups = [
        group["jail_served"].dropna()
        for _, group in df_c.groupby("obs_judge_name")
    ]

    if len(groups) >= 2:
        f_stat, p_val = stats.f_oneway(*groups)

        results.append({
            "county": county,
            "F_stat": f_stat,
            "p_value": p_val,
            "num_judges": len(groups)
        })

anova_results = pd.DataFrame(results)
anova_results
```

**Interpretation**

This table reports one-way ANOVA results testing whether jail outcomes differ across judges within each county.

```python
# Generate logistic regression results for a selected race baseline
def race_logit_results(baseline_race):
    df_clean = df_selected_newrace.copy()

    # Keep only the race groups used in the original figure
    df_clean = df_clean[df_clean["race"].isin(["Black", "Hispanic", "White"])]

    # Set baseline race
    race_order = [baseline_race] + [
        r for r in ["Black", "Hispanic", "White"] if r != baseline_race
    ]
    df_clean["race"] = pd.Categorical(df_clean["race"], categories=race_order)

    # Create dummy variables and add intercept
    race_dummies = pd.get_dummies(df_clean["race"], drop_first=True).astype(int)
    X = sm.add_constant(race_dummies)

    results = []

    for col in received_cols:

        # Skip outcomes with no variation
        if df_clean[col].nunique() < 2:
            continue

        y = df_clean[col].astype(int)

        # Fit logistic regression model
        model = sm.Logit(y, X).fit(disp=0)

        for race_var in race_dummies.columns:
            coef = model.params[race_var]
            pval = model.pvalues[race_var]

            results.append({
                "Penalty": col,
                "Race": race_var,
                "Coefficient": coef,
                "Odds Ratio": np.exp(coef),
                "p-value": pval
            })

    return pd.DataFrame(results)


# Store logistic regression results under different baseline groups
black_baseline = race_logit_results("Black")
white_baseline = race_logit_results("White")

# Keep only results with p-value less than 0.1
black_baseline = black_baseline[black_baseline["p-value"] < 0.1]
white_baseline = white_baseline[white_baseline["p-value"] < 0.1]

# Keep only the penalty outcomes shown in the original figure
keep_penalties = [
    "received_fine",
    "received_probation",
    "received_community_service",
    "received_detention",
    "received_DUI"
]

black_baseline = black_baseline[black_baseline["Penalty"].isin(keep_penalties)]
white_baseline = white_baseline[white_baseline["Penalty"].isin(keep_penalties)]


# P-value cut off
ALPHA = 0.05

# Create readable labels combining penalty type and race group
def make_labels(df):
    return [
        f"{row.Penalty.replace('received_', '')} ({row.Race})"
        for _, row in df.iterrows()
    ]

# Color bars by statistical significance
def bar_color(p):
    return "#A32D2D" if p < ALPHA else "#185FA5"

# Plot aspect ratio
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot title
fig.suptitle("Logistic regression — race effect on penalties (p < 0.1 shown)", fontsize=13)

# Create the bar plots
for row_idx, (df, baseline) in enumerate([
    (black_baseline, "Black"),
    (white_baseline, "White")
]):

    # Sort by OR only — p-value chart uses same order
    or_df = df.sort_values("Odds Ratio", ascending=True)
    labels = make_labels(or_df)
    cols = [bar_color(p) for p in or_df["p-value"]]

    ax1 = axes[row_idx, 0]
    ax2 = axes[row_idx, 1]

    # Odds ratios
    ax1.barh(labels, or_df["Odds Ratio"], color=cols, height=0.55)
    ax1.axvline(1.0, color="gray", linewidth=1.2, linestyle="--", label="OR = 1.0 (null)")
    ax1.set_xlabel("Odds ratio")
    ax1.set_title(f"Odds ratios (baseline: {baseline}, descending)")
    ax1.legend(fontsize=8)

    # P-values — same order as OR chart
    ax2.barh(labels, or_df["p-value"], color=cols, height=0.55)
    ax2.axvline(ALPHA, color="#A32D2D", linewidth=1.2, linestyle="--", label="α = 0.05")
    ax2.axvline(0.1, color="#BA7517", linewidth=1.0, linestyle=":", label="α = 0.10")
    ax2.set_xlabel("p-value")
    ax2.set_title(f"P-values (baseline: {baseline})")
    ax2.set_xlim(0, 0.15)
    ax2.legend(fontsize=8)

    # Annotate p-values
    for i, p in enumerate(or_df["p-value"]):
        ax2.text(max(p, 0.003) + 0.001, i, f"{p:.4f}", va="center", fontsize=8)

# Shared legend for significance
legend_patches = [
    mpatches.Patch(color="#A32D2D", label="p < 0.05 (significant)"),
    mpatches.Patch(color="#185FA5", label="p ≥ 0.05 (trend only)")
]

fig.legend(handles=legend_patches, loc="lower center", ncol=4,
           fontsize=9, bbox_to_anchor=(0.5, -0.02))

# Plot layout
plt.tight_layout()

# Show the plot
plt.show()
```

### Chi Square Tests For Race And County

```python
# Set significance level and chi-square critical value (df = 1)
ALPHA = 0.05
CHI_CRIT = 3.84

# Define the list of counties to analyze (keeps subplot order consistent)
county_order = [
    "Arapahoe", "Adams", "Boulder", "Broomfield", "Denver",
    "Douglas", "El Paso", "Jefferson", "Larimer", "Mesa",
    "Pueblo", "Teller", "Weld"
]

# Define all penalty indicator variables
penalty_cols = [
    "received_fine",
    "received_probation",
    "received_treatment",
    "received_jail",
    "received_community_service",
    "received_vip",
    "received_alcohol_eval",
    "received_restitution",
    "received_detention",
    "received_DUI",
    "received_work_release",
    "received_monitoring"
]

# Compute chi-square statistics
all_chi_results = []

for county_name in county_order:

    # Subset data for the current county
    subset = df_race_county_penalty_clean[
        df_race_county_penalty_clean["county"] == county_name
    ]

    for col in penalty_cols:

        # Skip penalty variables that do not exist in the dataset
        if col not in subset.columns:
            continue

        # Skip variables with no variation (chi-square requires at least two categories)
        if subset[col].nunique() < 2:
            continue

        # Construct contingency table: race vs. penalty outcome
        table = pd.crosstab(subset["race"], subset[col])

        # Ensure table is valid for chi-square test
        if table.shape[0] < 2 or table.shape[1] < 2:
            continue

        # Perform chi-square test
        chi2, p, dof, ex = chi2_contingency(table)

        # Store results
        all_chi_results.append({
            "County": county_name,
            "Penalty": col,
            "Chi2": chi2,
            "p-value": p
        })

# Convert results to DataFrame
all_chi_results_df = pd.DataFrame(all_chi_results)


# Convert results into plotting structure
# This matches the original "county_data" format used for visualization
county_data = {}

for county_name in county_order:

    temp = all_chi_results_df[
        all_chi_results_df["County"] == county_name
    ]

    county_data[county_name] = {
        "Penalty": list(temp["Penalty"]),
        "Chi2": list(temp["Chi2"]),
        "p-value": list(temp["p-value"])
    }


# Visualization
# Create subplot grid (one row per county, two panels)
fig, axes = plt.subplots(len(county_data), 2, figsize=(14, len(county_data)*3))
fig.suptitle(
    "Chi-square and p-value comparison across counties",
    fontsize=16,
    y=1.02
)

for i, (county, data) in enumerate(county_data.items()):

    # Sort penalties by chi-square value (ensures consistent ordering across panels)
    df = pd.DataFrame(data).sort_values("Chi2")

    # Color chi-square bars based on critical threshold
    chi_colors = ["#0C447C" if v > CHI_CRIT else "#85B7EB" for v in df["Chi2"]]

    # Color p-values based on statistical significance
    pv_colors = ["#A32D2D" if p < ALPHA else "#185FA5" for p in df["p-value"]]

    # Chi-square panel
    axes[i,0].barh(df["Penalty"], df["Chi2"], color=chi_colors)
    axes[i,0].axvline(CHI_CRIT, linestyle="--", color="gray")
    axes[i,0].set_title(county)

    # P-value panel (same ordering as chi-square)
    axes[i,1].barh(df["Penalty"], df["p-value"], color=pv_colors)
    axes[i,1].axvline(ALPHA, linestyle="--", color="red")
    axes[i,1].set_xlim(0,1)

# Adjust layout and display figure
plt.tight_layout()
plt.show()
```

### Tukey HSD Results — Log Fine Amounts by Race

```python
# Tukey HSD Pairwise Differences and P-values — Log Fine Amounts

# This figure compares log fine amounts across race groups using Tukey HSD.
# Tukey HSD is useful after group comparison because it shows which specific pairs differ.

# Keep only positive fine amounts before applying log transformation
df_fines = df_quantify[df_quantify["fines_actual"] > 0].copy()

# Create a new variable: log of fine amount
df_fines["log_fines"] = np.log(df_fines["fines_actual"])

# Run Tukey HSD to compare pairwise race differences in log fine amounts
# endog = outcome variable
# groups = race categories being compared
# alpha = significance threshold
tukey_fines = pairwise_tukeyhsd(
    endog=df_fines["log_fines"],
    groups=df_fines["race"],
    alpha=ALPHA
)

# Convert Tukey output into a dataframe for visualization
# This makes the Tukey output easier to clean, sort, and visualize.
tukey_df = pd.DataFrame(
    tukey_fines._results_table.data[1:],
    columns=tukey_fines._results_table.data[0]
)

# Convert numeric columns from object type to float for plotting
tukey_df["meandiff"] = tukey_df["meandiff"].astype(float)
tukey_df["p-adj"] = tukey_df["p-adj"].astype(float)

# Create readable comparison labels for the y-axis
tukey_df["Comparison"] = tukey_df["group1"].astype(str) + " vs " + tukey_df["group2"].astype(str)

# Sort by mean difference so both panels use the same order
tukey_df = tukey_df.sort_values("meandiff", ascending=True)

# Color bars by adjusted p-value significance
# Red indicates statistically significant pairwise differences.
# Blue indicates non-significant pairwise differences.
colors = [
    sig_color if p < ALPHA else nonsig_color
    for p in tukey_df["p-adj"]
]

# Create figure with two panels
# ax1 is used for mean differences.
# ax2 is used for adjusted p-values.
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 6))
fig.suptitle("Tukey HSD Results — Log Fine Amounts by Race", fontsize=13)

# Left panel: mean differences
# X-axis: mean difference in log fine amounts
# Y-axis: race group comparisons
ax1.barh(tukey_df["Comparison"], tukey_df["meandiff"], color=colors, height=0.6)

# Add a vertical reference line at 0.
# Values to the right mean the second group has a higher mean log fine amount than the first group.
# Values to the left mean the second group has a lower mean log fine amount than the first group.
ax1.axvline(0, color="gray", linestyle="--", linewidth=1.2)
ax1.set_title("Mean Differences")
ax1.set_xlabel("Mean Difference")

# Right panel: adjusted p-values in the same order
# X-axis: adjusted p-values from Tukey HSD
# Y-axis: same race pair comparisons as the left panel
ax2.barh(tukey_df["Comparison"], tukey_df["p-adj"], color=colors, height=0.6)

# Add a vertical reference line at alpha = 0.05.
# Pairwise comparisons below this cutoff are considered statistically significant.
ax2.axvline(ALPHA, color=sig_color, linestyle="--", linewidth=1.2, label="α = 0.05")

ax2.set_title("Adjusted P-values")
ax2.set_xlabel("p-adjusted")

# Set x-axis range so all adjusted p-values fit clearly in the plot
ax2.set_xlim(0, max(0.10, tukey_df["p-adj"].max() * 1.1))

# Add legend for the alpha reference line
ax2.legend(fontsize=9)

# Label each bar with the exact adjusted p-value
for i, p in enumerate(tukey_df["p-adj"]):
    ax2.text(max(p, 0.002) + 0.005, i, f"{p:.4f}", va="center", fontsize=8)

# Legend for significance colors
legend_handles = [
    mpatches.Patch(color=sig_color, label="p < 0.05"),
    mpatches.Patch(color=nonsig_color, label="p ≥ 0.05")
]
ax1.legend(handles=legend_handles, fontsize=9)

plt.tight_layout()
plt.show()
```

### Fine Rate Variation Visualization (Judge & County)

```python
# Run ANOVA within each county to test whether fine rates differ across judges
results_rate = []

for county in df_subset["county"].unique():

    df_c = df_subset[df_subset["county"] == county]

    # Keep only judges with sufficient cases
    valid = df_c["obs_judge_name"].value_counts()
    valid = valid[valid >= 20].index
    df_c = df_c[df_c["obs_judge_name"].isin(valid)]

    # Skip counties with fewer than two judges
    if df_c["obs_judge_name"].nunique() < 2:
        continue

    # Create binary fine indicator per case
    df_c = df_c.copy()
    df_c["received_fine_binary"] = (df_c["fines"] > 0).astype(int)

    # Create fine-rate groups by judge
    groups = [
        group["received_fine_binary"].dropna()
        for _, group in df_c.groupby("obs_judge_name")
    ]

    if len(groups) >= 2:
        f_stat, p_val = stats.f_oneway(*groups)

        results_rate.append({
            "county": county,
            "F_stat": f_stat,
            "p_value": p_val,
            "num_judges": len(groups)
        })

# Store ANOVA results in the same format used by the plot
anova_df = pd.DataFrame(results_rate)


# Significance threshold
ALPHA = 0.05

# Sort counties by F-statistic
anova_sorted = anova_df.sort_values("F_stat", ascending=True)

# Color mapping based on significance
colors = ["#A32D2D" if p < ALPHA else "#185FA5" for p in anova_sorted["p_value"]]

# Create side-by-side plots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
fig.suptitle("One-way ANOVA — fine rate variation by judge within each county", fontsize=13)

# Plot 1: F-statistics
ax1.barh(anova_sorted["county"], anova_sorted["F_stat"], color=colors, height=0.6)
ax1.set_xlabel("F-statistic")
ax1.set_title("F-statistic (descending)")

# Legend explaining color meaning
ax1.legend(handles=[
    mpatches.Patch(color="#A32D2D", label="p < 0.05 (significant)"),
    mpatches.Patch(color="#185FA5", label="p ≥ 0.05 (not significant)")
], fontsize=9)

# Annotate number of judges for context
for i, (f, n) in enumerate(zip(anova_sorted["F_stat"], anova_sorted["num_judges"])):
    ax1.text(f + 0.03, i, f"n={n} judges", va="center", fontsize=8, color="#73726c")

# Plot 2: p-values
ax2.barh(anova_sorted["county"], anova_sorted["p_value"], color=colors, height=0.6)

# Reference line for alpha = 0.05
ax2.axvline(ALPHA, color="#A32D2D", linewidth=1.2, linestyle="--", label="α = 0.05")
ax2.set_xlabel("p-value")
ax2.set_title("P-values")
ax2.set_xlim(0, 0.42)
ax2.legend(fontsize=9)

# Annotate exact p-values
for i, p in enumerate(anova_sorted["p_value"]):
    ax2.text(max(p, 0.005) + 0.005, i, f"{p:.4f}", va="center", fontsize=8)

# Final layout
plt.tight_layout()
# plt.savefig("anova_county_plot.png", dpi=150, bbox_inches="tight")
plt.show()
```

### Logistic Regression Visualization (Judge & County)

```python
# Prepare a clean dataset for judge-county severity modeling
# We start from a copy of the cleaned dataset to avoid modifying the original data.
df_jc = df_selected_new2.copy()

# Standardize county and judge names before creating judge-county groups
# Clean county and judge names to avoid duplicate categories caused by formatting issues
# (e.g., extra spaces, inconsistent capitalization, or typos)
df_jc["county"] = df_jc["county"].astype(str).str.strip().str.title()
df_jc["obs_judge_name"] = df_jc["obs_judge_name"].astype(str).str.strip()

# Fix known inconsistencies and data entry issues in county names
df_jc["county"] = df_jc["county"].replace({
    "Teller County": "Teller",
    "Denver, Teller": "Teller",
    "Ei Paso": "El Paso",
    'The Defendant Appeared In Person At The Larimer County Justice Center, But On The Docket It Said "Larimer County"': "Larimer"
})

# Filter usable observations
# Remove rows with missing key variables
df_jc = df_jc.dropna(subset=["county", "obs_judge_name", "penalty_severity"])

# Exclude "Unclassified" outcomes since they cannot be ranked in severity
df_jc = df_jc[df_jc["penalty_severity"] != "Unclassified"].copy()

# Convert ordered penalty categories into numeric severity scores
# Map ordered penalty categories into a numeric severity scale
# This allows us to treat the outcome as an ordinal variable
severity_map = {
    "Probation": 1,
    "Detention": 2,
    "Work Release": 3,
    "Jail": 4
}
df_jc["penalty_score"] = df_jc["penalty_severity"].map(severity_map)

# Combine judge and county to model each judge-county pair separately
df_jc["judge_county"] = df_jc["obs_judge_name"] + " | " + df_jc["county"]

# Keep only judge-county groups with enough observations for more stable estimates
# Count the number of observations for each judge-county pair
jc_counts = df_jc["judge_county"].value_counts()

# Keep only groups with at least 10 observations to improve stability
# (small groups can lead to noisy or unreliable estimates)
top_jc = jc_counts[jc_counts >= 10].index.tolist()

# Filter dataset to include only these groups
df_jc = df_jc[df_jc["judge_county"].isin(top_jc)].copy()
```

```python
# Set reference group as the largest group
jc_order = df_jc["judge_county"].value_counts().index.tolist()
df_jc["judge_county"] = pd.Categorical(df_jc["judge_county"], categories=jc_order, ordered=True)

# Create dummy variables for judge-county groups, dropping the reference group
X = pd.get_dummies(df_jc["judge_county"], drop_first=True)
X = X.astype(float)

# Use OrderedModel because penalty_score is an ordered severity outcome
model_jc = OrderedModel(
    df_jc["penalty_score"],
    X,
    distr="logit"
)

# Fit model using BFGS optimization
result_jc = model_jc.fit(method="bfgs", disp=False)
```

```python
# Build a results table from the ordinal logistic model
coef_df_jc = pd.DataFrame({
    "Variable": result_jc.params.index,
    "Coefficient": result_jc.params.values,
    "Odds Ratio": np.exp(result_jc.params.values), # convert log-odds to odds ratios
    "p-value": result_jc.pvalues.values
})

# Remove threshold parameters (not actual judge-county effects)
coef_df_jc = coef_df_jc[~coef_df_jc["Variable"].str.contains("/")].copy()

# Add case counts for each judge-county group to assess reliability
jc_case_counts = (
    df_jc["judge_county"]
    .value_counts()
    .rename_axis("Variable")
    .reset_index(name="cases")
)

coef_df_jc = coef_df_jc.merge(jc_case_counts, on="Variable", how="left")

# Keep only the most common groups to avoid unstable estimates（15）
coef_df_jc = coef_df_jc.sort_values("cases", ascending=False).head(15).copy()

# Sort for visualization (ordered by effect size)
coef_df_jc = coef_df_jc.sort_values("Odds Ratio", ascending=True)

# Format labels for better readability in the plot
coef_df_jc["Label"] = coef_df_jc["Variable"].str.replace(" | ", "\n", regex=False)

ALPHA = 0.05
sig_color = "#A32D2D"
nonsig_color = "#B8B5AC"

# Color by statistical significance
or_colors = [
    sig_color if p < ALPHA else nonsig_color
    for p in coef_df_jc["p-value"]
]

pv_colors = [
    sig_color if p < ALPHA else nonsig_color
    for p in coef_df_jc["p-value"]
]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 9))

fig.suptitle(
    "Ordered Logistic Regression — Penalty Severity by Judge & County",
    fontsize=14
)

# Left panel: Odds ratios (effect size relative to reference group)
ax1.barh(coef_df_jc["Label"], coef_df_jc["Odds Ratio"], color=or_colors)
ax1.axvline(1.0, color="gray", linestyle="--", linewidth=1.5, label="OR = 1 (null)")
ax1.set_title("Odds Ratios for Penalty Severity")
ax1.set_xlabel("Odds Ratio")

legend_elements = [
    Patch(facecolor=sig_color, label="Significant (p < 0.05)"),
    Patch(facecolor=nonsig_color, label="Not significant"),
]

ax1.legend(
    handles=legend_elements + [ax1.lines[0]],
    labels=[
        "Significant (p < 0.05)",
        "Not significant",
        "OR = 1 (null)"
    ],
    loc="upper right"
)

# Right panel: p-values (log scale to distinguish small values)
ax2.barh(coef_df_jc["Label"], coef_df_jc["p-value"], color=pv_colors)
ax2.axvline(ALPHA, color=sig_color, linestyle="--", linewidth=1.5, label=f"α = {ALPHA}")
ax2.set_title("Significance of Predictors")
ax2.set_xlabel("p-value (log scale)")
ax2.set_xscale("log")

ax2.legend(
    handles=legend_elements + [ax2.lines[0]],
    labels=[
        "Significant (p < 0.05)",
        "Not significant",
        f"α = {ALPHA}"
    ],
    loc="upper right"
)

plt.tight_layout()
plt.show()
```

```python
# Load data
county_sheets = [
    '2022 Adams', '2022 Arapahoe', '2022 Boulder', '2022 Broomfield',
    '2022 Denver', '2022 Douglas', '2022 El Paso', '2022 Jefferson',
    '2022 Larimer', '2022 Mesa', '2022 Pueblo', '2022 Weld'
]

judge_col  = 'Sentencing Judge (or jury trial judge if entire case was acquitted) '
charge_col = 'Original Charge #1'
fine_col   = 'Total fines ordered'

frames = []
for sheet in county_sheets:
    df = pd.read_excel(madd2022_rawdata, sheet_name=sheet, header=1)
    df = df[df['County'].notna()]
    frames.append(df[[judge_col, charge_col, fine_col, 'County']])

combined = pd.concat(frames, ignore_index=True)

# Clean data
combined['fine'] = pd.to_numeric(combined[fine_col], errors='coerce')
combined = combined[(combined['fine'] > 0) & (combined[judge_col].notna())].copy()
combined[judge_col] = combined[judge_col].str.strip()

# Remove top 5% outliers
CUTOFF    = combined['fine'].quantile(0.95)
n_before  = len(combined)
combined  = combined[combined['fine'] <= CUTOFF].copy()
n_removed = n_before - len(combined)

# Severity classification
def classify_severity(charge):
    c = str(charge).upper()
    if '3+' in c or 'FELONY' in c:   return '3+ Priors'
    if '2+' in c or '3RD' in c:      return '2+ Priors'
    if '1 PRIOR' in c or '2ND' in c: return '1 Prior'
    if 'DWAI' in c:                   return 'DWAI'
    if 'DUI' in c:                    return 'DUI (1st)'
    return 'Other'

combined['severity'] = combined[charge_col].apply(classify_severity)

# Fix judge name inconsistencies
name_fixes = {
    'Cheryl Ann Rowles-stokes': 'Cheryl Ann Rowles-Stokes',
    'Jonathon Martin\n': 'Jonathon Martin',
    '-': None,
}
combined[judge_col] = combined[judge_col].replace(name_fixes)
combined = combined[combined[judge_col].notna()].copy()

# Filter judges with enough cases
judge_counts = combined.groupby(judge_col)['fine'].count()
eligible     = judge_counts[judge_counts >= 10].index
df_plot      = combined[combined[judge_col].isin(eligible)].copy()

# Primary county per judge
judge_county = df_plot.groupby(judge_col)['County'].agg(
    lambda x: x.value_counts().index[0]
)

# Sort by median fine (descending)
judge_order = (
    df_plot.groupby(judge_col)['fine']
    .median()
    .sort_values(ascending=False)
    .index.tolist()
)

# Clean even split to avoid tick/label mismatch
n_judges = len(judge_order)
half     = n_judges // 2
col1     = judge_order[:half]
col2     = judge_order[half:]

print(f"col1: {len(col1)}, col2: {len(col2)}, total: {len(col1)+len(col2)}, judge_order: {len(judge_order)}")

# Severity config
severity_order = ['DUI (1st)', 'DWAI', '1 Prior', '2+ Priors', '3+ Priors', 'Other']
severity_colors = {
    'DUI (1st)': '#85B7EB',
    'DWAI':      '#0C447C',
    '1 Prior':   '#EF9F27',
    '2+ Priors': '#BA7517',
    '3+ Priors': '#A32D2D',
    'Other':     '#B4B2A9',
}

row_height = 1.2
box_width  = 0.45

# Drawing function
def draw_column(ax, judges_subset):
    for j_idx, judge in enumerate(judges_subset):
        judge_df = df_plot[df_plot[judge_col] == judge]
        y_center = j_idx * row_height

        if j_idx % 2 == 0:
            ax.axhspan(
                y_center - row_height / 2,
                y_center + row_height / 2,
                color='#f5f5f5', zorder=0
            )

        present_sevs = [s for s in severity_order if s in judge_df['severity'].values]
        n_sev = len(present_sevs)
        offsets = np.linspace(-row_height * 0.3, row_height * 0.3, n_sev) if n_sev > 1 else [0]

        for s_off, sev in zip(offsets, present_sevs):
            sev_data = judge_df[judge_df['severity'] == sev]['fine'].dropna()
            y_pos    = y_center + s_off
            color    = severity_colors[sev]

            if len(sev_data) >= 4:
                ax.boxplot(
                    sev_data,
                    positions=[y_pos],
                    vert=False,
                    widths=box_width,
                    patch_artist=True,
                    showfliers=False,
                    boxprops=dict(facecolor=color, alpha=0.75, linewidth=1),
                    medianprops=dict(color='black', linewidth=2),
                    whiskerprops=dict(color=color, linestyle='--'),
                    capprops=dict(color=color),
                    zorder=3,
                )
                mean_val = sev_data.mean()
                ax.plot(
                    [mean_val, mean_val],
                    [y_pos - box_width / 2, y_pos + box_width / 2],
                    color='red', linewidth=1.5, zorder=4
                )

    n = len(judges_subset)
    ticks  = [i * row_height for i in range(n)]
    labels = [f"{j}  [{judge_county[j]}]  n={judge_counts[j]}" for j in judges_subset]

    ax.set_yticks(ticks)
    ax.set_yticklabels(labels, fontsize=13)
    ax.set_ylim(-row_height, n * row_height)
    ax.invert_yaxis()
    ax.set_xlim(0, CUTOFF * 1.05)
    ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'${x:,.0f}'))
    ax.tick_params(axis='x', labelsize=12)
    ax.set_xlabel('Total fines ordered ($)', fontsize=14)
    ax.grid(axis='x', linestyle='--', alpha=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

# Create figure
fig_height = len(col1) * row_height + 4
fig, (ax1, ax2) = plt.subplots(
    1, 2,
    figsize=(26, fig_height),
    gridspec_kw={'wspace': 0.3}
)

draw_column(ax1, col1)
draw_column(ax2, col2)

# Title
fig.suptitle(
    f'Total fines per judge — 2022 Colorado DUI cases (all counties)\n'
    f'Split by original charge severity  |  top 5% outliers removed (>${CUTOFF:,.0f}, n={n_removed} cases)\n'
    f'Sorted by median fine (highest → left plot, continues lowest → right plot)',
    fontsize=13,
    y=0.99
)

# Figure-level legend
legend_patches = [
    mpatches.Patch(color=severity_colors[s], label=s, alpha=0.8)
    for s in severity_order
]
legend_patches.append(
    plt.Line2D([0], [0], color='red', linewidth=1.5, label='Mean')
)
fig.legend(
    handles=legend_patches,
    title='Original Charge #1',
    fontsize=10,
    title_fontsize=11,
    loc='upper left',
    ncol=len(legend_patches),
    bbox_to_anchor=(0.01, 0.97),
    framealpha=0.95
)

plt.subplots_adjust(top=0.93, bottom=0.05, left=0.15, right=0.97)
plt.show()

print(f"Done. {n_judges} judges, {n_removed} outliers removed (>${CUTOFF:,.0f}).")
```

# Limitation & Suggested Way to Further Research

**Limitations**

Our analysis has several limitations. First, we were unable to account for all of the charges each defendant faced. While we know that every defendant was charged with a DUI and the associated fine, we do not have visibility into any additional charges or contributing factors that may have influenced the outcome. As a result, we should be cautious about interpreting our analysis, as the true legal context surrounding each case may be more complex than what is reflected in our data.

An important limitation to note is the arithmetic operations on ordinal ranks. Ordinal scales conveys order but not distance, so treating rank differences as numerically equal, as ordinal logistic regression implicitly does, is a known limitation. However, Mann-Whitney U and Kruskal-Wallis tests are still statistically appropriate for ordinal data because they don't involve in spacing assumptions. This is also why we focus on non-ranked analysis and use ranked analysis as supplementary support for our project.

Another limitation to consider is our sample size in each group, for example when we analyze by comparing groups like Gender vs Female or White defendants vs Black defendants, our groups won't have the same sample size. This raises a concern for possible bias and other potential problems. Having unequal groups would affect how standard errors become larger, estimates becoming less stable and how large groups may dominate the smaller groups. However, these problems typically only occur when the smaller group has extremely low counts (less than 20 or 30 samples), which is not the case for our data, all our groups meet this minimum threshold.

While it may still be a concern, we tried to limit the concerns as much as possible by picking nonparametric methods and models that uses Maximum Likelihood Estimation (explained in "Why are our models reliable?" section). Generally, all of our methods wouldn't be affected by the difference in our groups sample sizes, with the exception of 2 models: OLS Regression and ANOVA table. For OLS, having unequal groups or skewed distribution could affect the calculation results which is why we applied log transformation when appropriate. For ANOVA, having unequal groups could distort Type I error rates* if one group has a much larger variance or is much bigger than another. To minimize this, we confirmed that no single group was drastically larger than the others, and we interpret ANOVA results alongside the heatmap rather than in isolation.

Additionally, for chi-square tests we verified that the expected cell counts met the minimum threshold of 5. This ensures those results remain valid despite any group size differences.
